```python
import pandas as pd

data = {
    "height": [165, 172, None, 180, 158, 175],
    "weight": [55, 70, 65, 80, 50, 68],
    "age": [25, 30, 28, 35, 22, 29],
    "gender": ["F", "M", "M", "M", "F", "M"],
}

df = pd.DataFrame(data)
```

```python
print(df.isnull().sum()) # 결측치 개수 
```

```python
# 결측치가 있는 행 제거
df2 = df.dropna()
```

```python
# 결측치 채우기 --> fillna()
# 숫자형 --> 평균값으로 채우기
df['height'].fillna(df['height'].mean())
```

```python
df3 = df.copy() # df복사 -> df3fh 
```

```python
# 인코딩 (Encoding)
# 문자 데이터를 숫자로 바꾸는 과정
```

```python
# Label Encoding (레이블 인코딩)
# 값에 숫자를 직접 매기는 것.
```

```python
df['gender_num'] = df['gender'].map({'F':0, 'M':1})
```

```python
df
```

```python
# One-Hot Encoding (원-핫 인코딩)
# True / False
# 1 / 0
```

```python
df_encoded = pd.get_dummies(df, columns=['gender'])
```

```python
df_encoded
```

```python
# 스케일링
# 숫자 값의 범위를 조정
# ex. 나이 --> 20~70, 연봉 --> 30,000,000
# (숫자를 비슷하게 조정해주는것)
```

```python
# MinMaScaler
# 사이킷런 안에 있다.
# 값을 0 ~ 1 범위로 바꾸어준다.
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['height_scaled']] = scaler.fit_transform(df[['height']])
```

```python
df
```

```python
# StandardScaler
# 평균이 0, 표준편차가 1이 되도록 조정
# 사이킷러 안에 있다.
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['height_standard']] = scaler.fit_transform(df[['height']])
```
